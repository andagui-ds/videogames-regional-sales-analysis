## Paso 2. Prepara los datos

Tratamiento y Justificación de Valores Ausentes

### 1. Transformación de Formatos e Imputación Estratégica

Al comenzar a trabajar con el Dataset, todas las columnas venían con
mayúsculas, por lo que se procedió a cambiar los nombres por completo a
minúsculas para estandarizar el código.

- **Columna `user_score`:** Se identificó la abreviatura **TBD** (*To Be
  Determined*). Dado que representa información que aún no existe, se
  reemplazó sistemáticamente por valores nulos legítimos (`NaN`),
  permitiendo transformar con éxito el tipo de dato de la columna de
  tipo genérico (`object`) a un tipo flotante analítico (`float64`).
- **Columna `year_of_release`:** Se modificó su tipo de dato de
  `float64` a un formato entero nativo (`Int64`) para representar
  correctamente los años cronológicos. Los valores ausentes en esta
  columna se imputaron cruzando la información basándose en otras
  plataformas del mismo juego; se utilizó el método `.transform('max')`
  agrupado por el nombre del juego (`name`) para **heredar el año de
  lanzamiento** si este se encontraba disponible en otra fila. Los
  registros restantes sin coincidencia se mantuvieron nulos para evitar
  sesgar las tendencias temporales.
- **Columna `rating`:** Los valores vacíos se rellenaron con la etiqueta
  categórica **“Unknown”**. Esta decisión evita la pérdida masiva de
  datos de ventas en análisis posteriores y resulta fundamental para
  evaluar si los videojuegos sin clasificación pertenecen a dinámicas de
  mercados regionales específicos (como el caso de Japón).

### 2. ¿Por qué existen valores ausentes en el Dataset?

La presencia de datos nulos en las columnas críticas de calificaciones
(`critic_score`, `user_score`) y clasificaciones (`rating`) obedece a
razones operacionales e históricas de la propia industria:

- **Anacronismo de Plataformas de Agregación:** Muchos de los
  videojuegos presentes en el dataset pertenecen a las décadas de **1890
  y 1990**. En aquella época no existían plataformas de recopilación
  masiva de reseñas web como *Metacritic*, por lo que las notas de la
  crítica y de los usuarios para esos títulos antiguos nunca existieron
  digitalmente.
- **Falta de Masa Crítica / Exclusividades:** Los títulos
  independientes, juegos de nicho o lanzamientos exclusivos de regiones
  muy específicas no lograron acumular el volumen mínimo necesario de
  análisis de prensa o votos de la comunidad digital requeridos para
  generar un promedio estadístico estable.
- **Sesgo Geográfico del Sistema ESRB:** El sistema de clasificación por
  edades **ESRB** nació formalmente en **1994** y su enfoque operativo
  está diseñado casi exclusivamente para el mercado de **América del
  Norte**. Por lo tanto, los videojuegos lanzados antes de ese año, o
  aquellos desarrollados de manera exclusiva para el mercado asiático
  (que influyen directamente en `jp_sales`), carecen por completo de
  esta certificación norteamericana.

### 5.0 Prueba las siguientes hipótesis

```python
# Aseguramos trabajar con el dataset filtrado del período reciente (2014-2016)
df_reciente = df[df['year_of_release'] >= 2014].copy()

# Definimos el umbral de significancia estadística alfa
alpha = 0.05

print("=== PRUEBAS DE HIPÓTESIS ESTADÍSTICAS ===\n")

# 1. Extraemos las calificaciones de usuarios para cada plataforma (eliminando nulos)
user_xbox = df_reciente[(df_reciente['platform'] == 'XOne') & (df_reciente['user_score'].notna())]['user_score']
user_pc = df_reciente[(df_reciente['platform'] == 'PC') & (df_reciente['user_score'].notna())]['user_score']

# 2. Ejecutamos la prueba t de Student para muestras independientes
# Usamos equal_var=False porque no asumimos que las varianzas de ambas poblaciones sean iguales (Prueba de Welch)
results = st.ttest_ind(user_xbox, user_pc, equal_var=False)

# 3. Mostramos los resultados obtenidos
print(f"Calificación promedio en Xbox One: {user_xbox.mean():.2f}")
print(f"Calificación promedio en PC:       {user_pc.mean():.2f}")
print(f"Valor p (p-value):                 {results.pvalue}\n")

# 4. Evaluamos la decisión estadística frente a nuestro umbral alfa
if results.pvalue < alpha:
    print("Resultado: Se rechaza la hipótesis nula (H0).")
    print("Conclusión: Las calificaciones promedio de los usuarios para Xbox One y PC SON diferentes.")
else:
    print("Resultado: No se puede rechazar la hipótesis nula (H0).")
    print("Conclusión: No hay evidencia estadística suficiente para decir que son diferentes. Las calificaciones promedio son las mismas.")
```

### Conclusión y Justificación (Prueba de Hipótesis 1)

**Resultado Estadístico:** \* **Valor p (*p-value*):** $0.1392$ (13.92%)
\* **Alfa ($\alpha$):** $0.05$ (5%)

**Justificación:** Dado que el valor p resultante ($0.1392$) es
**mayor** que nuestro nivel de significancia establecido
($\alpha = 0.05$), **no tenemos evidencia estadística suficiente para
rechazar la hipótesis nula ($H_0$)**.

En términos estadísticos, la diferencia observada entre la calificación
promedio de Xbox One (6.59) y la de PC (6.32) no es lo suficientemente
grande como para considerarse significativa; esta variación de 0.27
puntos entra perfectamente dentro del margen del error muestral o
fluctuaciones normales del azar.

**Conclusión:** Se acepta la premisa de la hipótesis nula. Podemos
afirmar con seguridad que **las calificaciones promedio de los usuarios
para las plataformas Xbox One y PC son las mismas** en el período
analizado (2014-2016). Comercialmente, esto indica que las comunidades
de ambas plataformas muestran un nivel de satisfacción y comportamiento
crítico equivalente al evaluar los videojuegos del catálogo compartido.

**Conclusión:** Podemos afirmar que **las calificaciones promedio de los
usuarios para Xbox One y PC son estadísticamente las mismas**. Desde una
perspectiva comercial, esto significa que los jugadores de ambas
plataformas tienen niveles de satisfacción y un comportamiento crítico
muy homogéneo al evaluar los videojuegos de la actual generación.

### Hipótesis 2: Calificaciones de Usuarios en los Géneros de Acción vs. Deportes

- **Resultado Métrico:** \* Calificación promedio Acción: **6.77**
  - Calificación promedio Deportes: **5.23**
  - Valor p (*p-value*): **$9.65 \times 10^{-15}$** (un valor
    prácticamente equivalente a $0$)

**Justificación:** Dado que el valor p es drásticamente **menor** que
nuestro umbral alfa ($0.05$), **se rechaza contundentemente la hipótesis
nula ($H_0$)** en favor de la hipótesis alternativa ($H_1$). Existe una
diferencia matemática masiva y real entre las puntuaciones de ambos
grupos que no puede explicarse por casualidad.

**Conclusión:** Es **completamente correcto afirmar que las
calificaciones promedio de los usuarios para Acción y Deportes son
diferentes**. En el mercado real, este fenómeno nos indica que el
público que consume videojuegos de deportes es significativamente más
estricto y tiende a calificar los títulos de forma más baja (promedio de
$5.23$) que aquellos que disfrutan de juegos de acción (promedio de
$6.77$).

### Explicaciónes finales del paso 5

### 1. Formulación de las Hipótesis Nula y Alternativa

Para realizar cualquier contraste estadístico, es necesario traducir las
preguntas de negocio en un par de afirmaciones matemáticas que sean
mutuamente excluyentes:

- **Hipótesis Nula ($H_0$):** Representa la afirmación del “no cambio”,
  la **igualdad** o la ausencia de efectos significativos. Esta
  hipótesis asume de partida que las diferencias observadas entre los
  grupos se deben puramente al azar o a fluctuaciones del muestreo.
  **Siempre contiene el signo de igualdad ($=$).**
- **Hipótesis Alternativa ($H_1$):** Es la conjetura que rompe el estado
  basal; afirma que **sí existe una diferencia o efecto real** y
  significativo entre las poblaciones analizadas. **Siempre contiene un
  signo de desigualdad ($\neq$).**

#### Aplicación en el Proyecto:

- **Prueba 1 (Plataformas):** Evaluamos si las calificaciones promedio
  de los usuarios para Xbox One y PC son las mismas.
  - $H_0: \mu_{\text{Xbox One}} = \mu_{\text{PC}}$ (Las calificaciones
    promedio son iguales).
  - $H_1: \mu_{\text{Xbox One}} \neq \mu_{\text{PC}}$ (Las
    calificaciones promedio son diferentes).
- **Prueba 2 (Géneros):** Evaluamos si las calificaciones promedio de
  los usuarios para los géneros de Acción y Deportes son diferentes.
  - $H_0: \mu_{\text{Acción}} = \mu_{\text{Deportes}}$ (Las
    calificaciones promedio son iguales).
  - $H_1: \mu_{\text{Acción}} \neq \mu_{\text{Deportes}}$ (Las
    calificaciones promedio son diferentes).

### 2. Criterio Utilizado para Probar las Hipótesis y Justificación

El criterio matemático seleccionado fue la **Prueba $t$ de Student para
dos muestras independientes**, ejecutada mediante la función
`scipy.stats.ttest_ind()` configurada con el parámetro `equal_var=False`
(Prueba de Welch).

#### ¿Por qué es el criterio correcto?

1.  **Tipo de Variables:** Estamos comparando una variable cuantitativa
    continua (las puntuaciones de satisfacción de los usuarios,
    `user_score`) dividida a través de grupos categóricos mutuamente
    excluyentes (o el juego es de PC o es de Xbox One; o pertenece a
    Acción o a Deportes).
2.  **Muestras Independientes:** Las observaciones de un grupo no
    influyen ni están relacionadas de manera emparejada con las del otro
    (las calificaciones de los usuarios de PC son independientes de las
    de consola).
3.  **Justificación de `equal_var=False`:** En el dataset real, el
    volumen de registros difiere sustancialmente entre grupos (por
    ejemplo, hay una cantidad masiva de juegos de Acción en comparación
    con los de Deportes). No asumir varianzas poblacionales iguales
    (aplicar la corrección de Welch) le otorga robustez a la prueba y
    disminuye drásticamente la probabilidad de cometer un error de Tipo
    I (falso positivo).

#### Regla de Decisión con el Valor p (p-value):

Establecimos un umbral de significancia estándar de **$\alpha = 0.05$**.
\* Si $\text{valor p} < 0.05$: La probabilidad de que la diferencia
observada sea obra del azar es menor al 5%. Por ende, **se rechaza
$H_0$** en favor de $H_1$. \* Si $\text{valor p} \geq 0.05$: No hay
fuerza estadística suficiente para dudar del estado basal. Por ende,
**no se puede rechazar $H_0$**.

### 1. Síntesis de la Preparación de los Datos

El dataset original presentaba anomalías críticas que nublaban el
análisis. Se homogeneizaron los nombres de las columnas a minúsculas y
se detectó que el string `'tbd'` (*To Be Determined*) en la columna
`user_score` estaba secuestrando el tipo de dato, forzándolo a texto. Se
resolvió de manera estadística transformándolo en un nulo legítimo
(`NaN`) para permitir la conversión a flotante. Asimismo, la columna
`year_of_release` se optimizó a un tipo entero nativo (`Int64`),
logrando rescatar 123 registros nulos mediante el mapeo cruzado de
plataformas del mismo título. Las ausencias masivas en calificaciones se
conservaron intactas para preservar la varianza natural del mercado.

### 2. Dinámica del Mercado y Ciclo de Vida

La industria experimentó su máximo esplendor histórico entre 2005 y
2011. Sin embargo, los datos demuestran que el mercado es cíclico y
dependiente del hardware: \* Las plataformas tienen un ciclo de vida
útil promedio de **10 a 11 años** desde su lanzamiento hasta su
desaparición absoluta. \* Una consola tarda entre **3 y 5 años** en
alcanzar su madurez comercial y su pico de facturación. \* Para
construir el modelo predictivo de 2017, se determinó que **solo los
datos de 2014 a 2016 son relevantes**, dado que capturan la estabilidad
de la octava generación de consolas, dejando fuera plataformas ya
obsoletas.

### 3. Perfil Comercial y Potencial de Rentabilidad para 2017

El mercado global está dominado por la **PlayStation 4** y la **Xbox
One**, acompañadas por la **PC** como una plataforma de ingresos
estables y linealmente predecibles. El análisis de diagramas de caja
reveló que, aunque la mayoría de los juegos individuales venden menos de
500,000 unidades, la rentabilidad de las plataformas líderes está
impulsada por éxitos atípicos (*outliers*) masivos (juegos AAA).

Las reseñas de la crítica profesional muestran una **correlación
positiva moderada (~0.40)** con el éxito financiero de los títulos,
consolidándose como un factor predictivo útil. Por el contrario, las
calificaciones de los usuarios no muestran relación con las ventas
finales, sugiriendo que las decisiones de compra masivas están más
influenciadas por el marketing y la lealtad de marca que por la opinión
del consumidor digital.

### 4. Segmentación Regional Estratégica

Para maximizar la inversión publicitaria y la distribución en 2017, se
deben seguir tres enfoques completamente diferenciados por región: \*
**Norteamérica (NA):** Es un mercado de alta competencia. El enfoque
debe ser biconsola (**PS4 y Xbox One**). Los géneros prioritarios son
**Shooter** y **Action**, apuntando principalmente a clasificaciones **M
(Mature)** de la ESRB. \* **Europa (UE):** Presenta un monopolio de
consumo a favor de Sony (**PS4**). La distribución debe priorizar esta
plataforma con títulos de **Action, Shooter y Sports** (con especial
énfasis en simuladores de fútbol). \* **Japón (JP):** Es un ecosistema
aislado que ignora la tendencia occidental de consolas de sobremesa y
clasificaciones americanas (ESRB). La inversión aquí debe dirigirse
exclusivamente a la consola portátil **Nintendo 3DS** y al desarrollo
del género **Role-Playing (RPG)**.

A través del contraste de hipótesis, se validó con seguridad estadística
que **los usuarios de Xbox One y PC comparten los mismos niveles de
satisfacción promedio** en sus puntuaciones, abriendo la puerta a
estrategias de marketing unificadas para comunidades *cross-play*. Por
otra parte, se confirmó que **las calificaciones del género de Acción y
Deportes son significativamente diferentes**, siendo los usuarios de
juegos deportivos los más severos y críticos con sus valoraciones
anuales.
